```python
import pandas as pd
```

```python
# initialize list for data
gameList = []

# strings for start and end of wrapping for datafram
firstString = "data-ds-itemkey=\\\"App_"
secondString = "\\\" data-ds-tagids"

# filelist for woke games
fileList = [ "Data/Is-It-Woke-Not-Recommended.json",
             "Data/Woke-and-Censored-Games-Alert-Not-Recommended.json",
             "Data/Woke-Content-Detector-Not-Recommended.json"]

for fileName in fileList:
    # open files
    file = open(fileName, encoding="utf-8")
    string = file.read()

    # flag for has finished going through entire file
    flag = True
    while flag:
        if string.find(firstString) == -1:
            flag = False
            break
        # find and append appids to list
        gameList.append(int(string[string.find(firstString)+len(firstString):string.find(secondString)]))

        # get rid of previous data
        string = string[string.find(secondString)+len(secondString):len(string)]


# use a set to get rid of duplicate values
noDuplicates = list(set(gameList))

# use Pandas series to print to a csv file for later usage
array = pd.Series(noDuplicates)
array.to_csv("Data/wokegames.csv")
```
